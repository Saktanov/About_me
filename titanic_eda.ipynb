```python
# Импортируем библиотеки для анализа данных
import pandas as pd
import matplotlib.pyplot as plt

print("Окружение готово!")
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

print("Данные успешно загружены!")
print("Размер таблицы:", df.shape)
```

```python
df.head()
```

```python
df.dtypes
```

```python
print("Количество пропусков в каждом столбце:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
df.describe(include="all").T
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["age"].dropna(), bins=20, edgecolor="black")

plt.title("Распределение возраста пассажиров")
plt.xlabel("Возраст")
plt.ylabel("Количество пассажиров")

plt.show()
```

```python
df["sex"].value_counts().plot(
    kind="bar",
    figsize=(7, 5),
    edgecolor="black"
)

plt.title("Количество пассажиров по полу")
plt.xlabel("Пол")
plt.ylabel("Количество пассажиров")
plt.xticks(rotation=0)

plt.show()
```

```python
print("Доля выживших по полу:")
print(df.groupby("sex")["survived"].mean().round(3))

print("\nКоличество пассажиров по классам:")
print(df["class"].value_counts())

print("\nКоличество пропусков в возрасте:")
print(df["age"].isnull().sum())
```

Наблюдения по результатам анализа

Пропущенные данные. В столбце age есть пропущенные значения. Поэтому при анализе возраста необходимо учитывать, что информация доступна не для всех пассажиров.

Различия в выживаемости. Доля выживших отличается у мужчин и женщин. Это можно заметить при сравнении средних значений столбца survived по полу.

Распределение пассажиров по классам. Пассажиры распределены по классам билетов неравномерно. Это показывает, что состав пассажиров различался по классу билета.

Вывод: первичный анализ помог изучить структуру данных, выявить пропуски, сравнить категории и увидеть распределение возраста пассажиров.
